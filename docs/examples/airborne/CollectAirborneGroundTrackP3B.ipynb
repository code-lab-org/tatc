# Collect Airborne Ground Track: P-3B

This example demonstrates the TAT-C `collect_airborne_ground_track` analysis function:
given an aircraft (`Aircraft`) that flies a trajectory (`Trajectory`: its position,
direction, roll, and pitch over time) and carries an instrument (`PointedInstrument`),
compute the geometric ground footprint at each time.

1. **Plan** flight lines and their connecting paths with HyPlan.
2. **Sample** that same plan once per second (`tatc.preprocess.sample_flight_plan`) and
   **estimate** roll and pitch from the track and altitude changes
   (`tatc.preprocess.estimate_attitude`). The sampled positions follow HyPlan's geometry.
3. **Collect** the footprints with TAT-C, check their geometry, observe a level above the
   ground, and visualize the swath.

This is an illustrative planned flight over Arizona, not a recorded CAMP2Ex flight.
The estimated attitude is a modeling assumption, not measured aircraft attitude: a flight
recording in the IWG1 format gives the measured attitude instead
(`tatc.preprocess.states_from_iwg1`). The formula checks here test the geometry of the
calculation, not real-flight accuracy.

TAT-C reads a trajectory table and does not import HyPlan: the preprocessing functions
read HyPlan's outputs. Running this example requires HyPlan in addition to the example
dependencies: `pip install tatc[airborne]`.

```python
from pathlib import Path
import tempfile

import cartopy.crs as ccrs
import cartopy.feature as cfeature
from cartopy.io.img_tiles import GoogleTiles
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
from matplotlib.patches import Patch
import numpy as np
import pandas as pd
import shapely
from pyproj import Geod

from tatc.analysis import collect_airborne_ground_track, compute_airborne_ground_track
from tatc.preprocess import estimate_attitude, instrument_from_sensor, sample_flight_plan
from tatc.schemas import Aircraft, PointedInstrument, Trajectory
from tatc.utils import compute_geoid_height

GEOD = Geod(ellps="WGS84")
```

## Plan the Flight Lines with HyPlan

Use the HyPlan NASA P-3 model to plan five parallel 50 km flight lines, 10 km apart,
at 5.8 km altitude over central Arizona. These dimensions were inspired by the
CAMP2Ex P-3B survey legs; the location, date, and flight plan are illustrative.

HyPlan generates both the flight lines and the connecting turns. At the planned
speed of about 170 m/s, the model's 25-degree bank corresponds to a turn diameter
of about 12.6 km. Since the lines are only 10 km apart, a connector can bend away
before turning back toward the next line. This notebook retains that geometry.

The lead-in and lead-out are each 1 nautical mile. HyPlan's altitudes are above mean
sea level, whereas TAT-C's are above the WGS 84 ellipsoid: the EGM2008 geoid, about 30 m
below the ellipsoid near Phoenix (`tatc.utils.compute_geoid_height`), converts one to
the other.

The sensor is a HyPlan sensor model, converted to a TAT-C instrument by
`instrument_from_sensor`: a scanning sensor (a line scanner such as HyTES, AVIRIS-3, or
MASTER, a scanning lidar such as LVIS, or a side-looking radar such as UAVSAR) scans across
the track between the angles of its swath edges, and a frame camera views a tilted
rectangle. HyTES is selected as a line-scanner example; this is not a claim that this
aircraft/sensor configuration was flown in CAMP2Ex. At its 25-degree half angle the level
swath is about 5.4 km wide, so the 10 km line spacing leaves coverage gaps.

```python
import hyplan
from hyplan import FlightLine, compute_flight_plan, ureg
from importlib.metadata import version

print(f"HyPlan {version('hyplan')}, TAT-C {version('tatc')}")

# The aircraft and the flight: the values that differ from one aircraft to another
AIRCRAFT_NAME = "NASA P-3B"
HYPLAN_AIRCRAFT = hyplan.NASA_P3
ALTITUDE_M = 5800  # above mean sea level
NUMBER_OF_LINES = 5
LINE_LENGTH_KM = 50
LINE_SPACING_KM = 10
# Smoothing used only for attitude estimation; it does not change the path.
ATTITUDE_SMOOTH = pd.Timedelta(seconds=15)
# The sensor registered in HyPlan (for example "AVIRIS3", "HyTES", "MASTER", "LVIS", or
# "UAVSAR_Lband"). To explore another scan angle, set CUSTOM_HALF_ANGLE_DEG: the sensor
# is then a virtual setting, marked as such in the results and on the maps.
SENSOR = "HyTES"
CUSTOM_HALF_ANGLE_DEG = None  # for example 30.0 for a virtual half angle of 30 degrees
MAP_ZOOM, TURN_MAP_ZOOM = 11, 12  # the detail of the maps below

sensor = hyplan.create_sensor(SENSOR)
scanner = instrument_from_sensor(sensor)
if CUSTOM_HALF_ANGLE_DEG is not None:
    scanner = scanner.model_copy(
        update={
            "name": f"{sensor.name} (virtual, {CUSTOM_HALF_ANGLE_DEG:.1f} deg)",
            "cross_track_field_of_view": 2 * CUSTOM_HALF_ANGLE_DEG,
        }
    )
HALF_ANGLE_DEG = scanner.cross_track_field_of_view / 2
SENSOR_LABEL = f"{scanner.name}, half angle {HALF_ANGLE_DEG:.1f}°" + (
    "" if CUSTOM_HALF_ANGLE_DEG is None else " (VIRTUAL)"
)
print(
    f"{scanner.name}: cross-track field of view {scanner.cross_track_field_of_view:.1f} deg, "
    f"roll angle {scanner.roll_angle:.1f} deg, swept scan: {scanner.is_swept}"
)
print(
    f"swath width of a level scan at {ALTITUDE_M / 1000:.1f} km: "
    f"{2 * ALTITUDE_M * np.tan(np.radians(HALF_ANGLE_DEG)) / 1000:.1f} km"
)

center_latitude, center_longitude = 33.4, -111.9
takeoff_time = pd.Timestamp("2026-10-15 17:00:00", tz="UTC")

lines = []
for i in range(NUMBER_OF_LINES):
    longitude, latitude, _ = GEOD.fwd(
        center_longitude,
        center_latitude,
        90.0,
        (i - (NUMBER_OF_LINES - 1) / 2) * LINE_SPACING_KM * 1000.0,
    )
    lines.append(
        FlightLine.center_length_azimuth(
            lat=latitude,
            lon=longitude,
            length=LINE_LENGTH_KM * ureg.km,
            az=0.0 if i % 2 == 0 else 180.0,
            altitude_msl=ALTITUDE_M * ureg.meter,
            site_name=f"Line {i + 1}",
        )
    )

plan = compute_flight_plan(
    HYPLAN_AIRCRAFT(),
    lines,
    takeoff_time=takeoff_time.to_pydatetime(),
    n_samples=4000,
    start_offset=1,
    end_offset=1,
)
plan[
    [
        "segment_type",
        "segment_name",
        "distance",
        "time_to_segment",
        "start_altitude",
        "end_altitude",
    ]
]
```

## Sample the HyPlan Path and Estimate Attitude

`sample_flight_plan` samples each HyPlan segment at one-second intervals along its path
(including its connecting turns), at a constant ground speed within each segment, and
converts its altitudes to the ellipsoid with the EGM2008 geoid height at each state. The heading is the track
plus the crab angle that HyPlan computes for a flight line in a wind; this plan has no
wind, so the heading is the track.

The flight-plan output does not supply time-resolved roll and pitch. `estimate_attitude`
estimates roll as `atan(V * d(track)/dt / g)`, using ground speed `V`, the coordinated-turn
relation in still air, and pitch as `atan2(vertical_speed, V)`, a flight-path-angle
approximation without angle of attack. A centered moving average of `ATTITUDE_SMOOTH` is
applied twice to the turn and climb rates before calculating the angles. This reduces
sampling fluctuations but can blur short transitions; the smoothing is not calibrated to
a P-3B roll response, and it changes the estimated attitude only, not the positions.

TAT-C reads the resulting table into a `Trajectory` (with `Trajectory.from_dataframe`):
`time`, `latitude`, `longitude`, and `altitude`, and optionally `track`, `ground_speed`,
`heading`, `roll`, and `pitch`. A flight recording or another source of the same columns
can be read the same way.

```python
trajectory = estimate_attitude(
    sample_flight_plan(plan, takeoff_time),
    smoothing=ATTITUDE_SMOOTH,
)
turn_segments = list(plan.loc[plan["segment_type"] == "transit", "segment_name"])
SCENARIO_LABEL = "HyPlan path, estimated roll/pitch"
SETTINGS_LABEL = f"{SENSOR_LABEL} | {SCENARIO_LABEL}"

duration = trajectory["time"].iloc[-1] - trajectory["time"].iloc[0]
print(f"{len(trajectory)} samples over {duration}; {SCENARIO_LABEL}")
print(f"largest estimated bank angle: {trajectory['roll'].abs().max():.1f} degrees")
print(f"geoid height at the start: {compute_geoid_height(*trajectory[['longitude', 'latitude']].iloc[0]):.1f} m")
first_turn = np.flatnonzero(trajectory["segment_name"].eq(turn_segments[0]))
sample_rows = [0, first_turn[0] // 2, first_turn[0] + 15, first_turn[0] + 100]
trajectory[
    ["time", "latitude", "longitude", "altitude", "track", "ground_speed", "roll", "pitch"]
].iloc[sample_rows]
```

## Collect the Footprints with TAT-C

The sensor is a `PointedInstrument` that scans across the track (`view_geometry="scan"`) with a cross-track field of view of twice the half angle (`HALF_ANGLE_DEG`; it is a different quantity from the bank angle of the aircraft in a turn). Its view sweeps along the track (`is_swept=True`), so its footprint at each time is the strip that its scan line sweeps over the one-second time step centered on that time. The sensor is carried by an `Aircraft` that flies the trajectory: the roll, pitch, and heading of the trajectory rotate the scan with the aircraft's body, as the attitude of a satellite rotates the views of its instruments (an aircraft's roll, right wing down positive, tilts the scan to the left, as an instrument's positive roll angle does).

`collect_airborne_ground_track` intersects the rotated scan rays with the WGS 84 ellipsoid, as `collect_ground_track` does for a satellite, so that banked turns and the curvature of the Earth are modeled. Samples that cannot observe the ground are flagged with `valid_obs = False` and have an empty footprint, for example below an observed level (see below). With `solar_altaz=True`, it also gives the sun's altitude and azimuth and the glint angle at the center of each footprint.

The footprints are the geometric area that the scan covers on the WGS 84 ellipsoid. They do not account for the terrain, clouds, the sensitivity of the sensor, or the pointing limits of its mount, so they show where the sensor can see in principle and not what it will observe.


The maps below use Cartopy, as in the other TAT-C examples, with the
[USGS Topo basemap](https://basemap.nationalmap.gov/arcgis/rest/services/USGSTopo/MapServer)
to show roads, place names, and shaded terrain. The first rendering needs an internet
connection; map tiles are cached in a temporary directory. Set `USE_TOPO_BASEMAP = False`
to draw the maps without a background image, which downloads nothing.
**The basemap provides geographic context only: these footprints are projected onto
the WGS 84 ellipsoid, not onto the terrain shown on the map.**

```python
aircraft = Aircraft(
    name=AIRCRAFT_NAME,
    trajectory=Trajectory.from_dataframe(trajectory),
    instruments=[scanner],
)
footprints = collect_airborne_ground_track(aircraft, solar_altaz=True)
print(f"{len(footprints)} footprints, {int(footprints['valid_obs'].sum())} valid")
print(
    f"solar altitude {footprints['solar_alt'].min():.0f} to "
    f"{footprints['solar_alt'].max():.0f} degrees"
)
print(SETTINGS_LABEL)
footprints.drop(columns="geometry").iloc[sample_rows]
```

## Check the Footprints

The checks compare the footprints with formulas that do not use the analysis code, so they test the analysis function whatever the trajectory is. The scan edges are the centers of the views of two instruments mounted at the roll angles of the edges, `c + a` and `c - a` for a half angle `a` and the scan's center roll angle `c` (zero for a symmetric scanner; a positive roll angle of an instrument looks toward the left wing). For a height `h`, an aircraft roll `φ`, and a pitch `θ`:

1. On a straight, level leg, the distance between the two scan edges is the swath width `h (tan(c + a) - tan(c - a))`. This is the formula for a flat surface: the ellipsoid adds the curvature of the Earth, a fraction of a percent here.
2. On a straight, level leg, the one-second footprints tile the swath: the area of their union is the swath width times the distance flown.
3. In a banked turn, the center of the swath, midway between its scan edges, moves across the track by `-h (tan(c + a + φ) + tan(c - a + φ)) / 2 / cos(θ)`, to the right if positive.

These checks test the geometry of the calculation, not the footprints against the observations of an instrument.

```python
# the scan edges: the view centers of instruments mounted at the edge angles
edge_angles = (scanner.roll_angle + HALF_ANGLE_DEG, scanner.roll_angle - HALF_ANGLE_DEG)
edges = Aircraft(
    name=AIRCRAFT_NAME,
    trajectory=aircraft.trajectory,
    instruments=[
        PointedInstrument(
            cross_track_field_of_view=1, along_track_field_of_view=1, roll_angle=angle
        )
        for angle in edge_angles  # left, then right
    ],
)
track = aircraft.get_track()


def get_edges(elevation=0):
    """Longitudes and latitudes (degrees) of the left and right scan edges."""
    centers = [
        edges.get_instrument(i).compute_footprint_center(track, elevation) for i in (0, 1)
    ]
    return (
        np.stack([center.longitude.degrees for center in centers], axis=1),
        np.stack([center.latitude.degrees for center in centers], axis=1),
    )


edge_longitude, edge_latitude = get_edges()


def cross_track_km(edge):
    """Distance (km) of a scan edge to the right of the track (negative to the left)."""
    azimuth, _, distance = GEOD.inv(
        trajectory["longitude"].to_numpy(),
        trajectory["latitude"].to_numpy(),
        edge_longitude[:, edge],
        edge_latitude[:, edge],
    )
    return distance * np.sin(np.radians(azimuth - trajectory["track"].to_numpy())) / 1000


left, right = cross_track_km(0), cross_track_km(1)
projected_crs = footprints.estimate_utm_crs()
center, half_angle = np.radians(scanner.roll_angle), np.radians(HALF_ANGLE_DEG)
rows = {}

# 1 and 2: the interior of a straight, level leg
leg = np.flatnonzero(trajectory["segment_name"].eq("Line 3").to_numpy())[30:-30]
height = trajectory["altitude"].to_numpy()[leg].mean()
width_expected = height * (np.tan(center + half_angle) - np.tan(center - half_angle)) / 1000
width_computed = np.median((right - left)[leg])
rows["1. scan edge distance on a level leg (km)"] = (width_expected, width_computed)
leg_area = shapely.union_all(
    footprints.iloc[leg].to_crs(projected_crs).geometry.to_numpy()
).area / 1e6
leg_distance = GEOD.inv(
    trajectory["longitude"].iloc[leg[0]],
    trajectory["latitude"].iloc[leg[0]],
    trajectory["longitude"].iloc[leg[-1] + 1],
    trajectory["latitude"].iloc[leg[-1] + 1],
)[2] / 1000
rows["2. swept area of a level leg (km²)"] = (width_computed * leg_distance, leg_area)

# 3: the center of the swath in the turns
roll, pitch = np.radians(trajectory["roll"].to_numpy()), np.radians(trajectory["pitch"].to_numpy())
banked = (np.abs(trajectory["roll"].to_numpy()) > 5) & footprints["valid_obs"].to_numpy()
shift_expected = (
    -trajectory["altitude"].to_numpy()
    * (np.tan(center + half_angle + roll) + np.tan(center - half_angle + roll))
    / 2
    / np.cos(pitch)
    / 1000
)
shift_computed = (left + right) / 2
rows["3. center of the swath in turns, median (km)"] = (
    np.median(shift_expected[banked]),
    np.median(shift_computed[banked]),
)
largest_shift_difference = np.abs(shift_computed - shift_expected)[banked].max() * 1000

checks = pd.DataFrame(rows, index=["expected", "computed"]).T
checks["difference (%)"] = 100 * (checks["computed"] / checks["expected"] - 1)
print(
    f"{int(banked.sum())} samples in turns: largest difference of the center of the swath "
    f"from the formula {largest_shift_difference:.0f} m"
)
print(SETTINGS_LABEL)
checks.round(3)
```

```python
USE_TOPO_BASEMAP = True
# Map resolution (dots per inch); a lower value keeps the notebook small.
MAP_DPI = 80
GEOGRAPHIC = ccrs.PlateCarree()
TOPO_URL = (
    "https://basemap.nationalmap.gov/arcgis/rest/services/"
    "USGSTopo/MapServer/tile/{z}/{y}/{x}"
)
TILE_CACHE = Path(tempfile.gettempdir()) / "tatc-usgs-topo"
PHOENIX = (-112.0740, 33.4484)


def map_extent(*frames, padding=0.1):
    """Bounds in degrees, padded to keep the swath away from the map edge."""
    bounds = np.array([frame.total_bounds for frame in frames])
    west, south = bounds[:, :2].min(axis=0)
    east, north = bounds[:, 2:].max(axis=0)
    dx, dy = padding * (east - west), padding * (north - south)
    west, east, south, north = west - dx, east + dx, south - dy, north + dy
    # Leave enough east-west context around the long, narrow flight plan.
    middle = (west + east) / 2
    width = max(east - west, (north - south) / np.cos(np.radians((north + south) / 2)))
    return [middle - width / 2, middle + width / 2, south, north]


def draw_line_context(ax):
    """Roads, rivers, and urban areas of Natural Earth as a line drawing."""
    for category, name, color, width in (
        ("cultural", "urban_areas", "#b0b0b0", 0.8),
        ("physical", "rivers_north_america", "#7aa6c2", 0.9),
        ("cultural", "roads_north_america", "#909090", 0.6),
    ):
        feature = cfeature.NaturalEarthFeature(category, name, "10m")
        ax.add_feature(feature, facecolor="none", edgecolor=color, linewidth=width, zorder=1)
    return "Lines: Natural Earth roads, rivers, and urban areas"


def make_map(extent, title, zoom, background="topo"):
    """
    A Cartopy map with a USGS topographic background ("topo", if USE_TOPO_BASEMAP), a
    line drawing of roads, rivers, and urban areas ("lines"), or no background (None).
    """
    tiles = (
        GoogleTiles(url=TOPO_URL, cache=TILE_CACHE)
        if background == "topo" and USE_TOPO_BASEMAP
        else None
    )
    projection = tiles.crs if tiles is not None else ccrs.Mercator()
    fig, ax = plt.subplots(figsize=(8, 8), dpi=MAP_DPI, subplot_kw={"projection": projection})
    ax.set_extent(extent, crs=GEOGRAPHIC)
    if tiles is not None:
        ax.add_image(tiles, zoom, zorder=0)
        credit = "Basemap: USGS The National Map"
    elif background == "lines":
        credit = draw_line_context(ax)
    else:
        ax.set_facecolor("#f3f3f3")
        credit = "No basemap image (USE_TOPO_BASEMAP = False)"
    grid = ax.gridlines(
        draw_labels=["bottom", "left"], linewidth=0.4, color="gray", alpha=0.5
    )
    grid.xlabel_style = grid.ylabel_style = {"size": 9}
    ax.text(
        0.01,
        0.015,
        credit,
        transform=ax.transAxes,
        fontsize=8,
        zorder=10,
        bbox={"facecolor": "white", "edgecolor": "none", "alpha": 0.85, "pad": 3},
    )
    ax.set_title(title, fontsize=13, pad=26)
    ax.text(
        0.5,
        1.012,
        SETTINGS_LABEL,
        transform=ax.transAxes,
        ha="center",
        va="bottom",
        fontsize=8,
    )
    ax.plot(*PHOENIX, "ko", markersize=3, transform=GEOGRAPHIC, zorder=5, clip_on=True)
    ax.text(
        PHOENIX[0] + 0.015,
        PHOENIX[1] + 0.015,
        "Phoenix",
        transform=GEOGRAPHIC,
        fontsize=10,
        weight="bold",
        zorder=6,
        clip_on=True,
        bbox={"facecolor": "white", "edgecolor": "none", "alpha": 0.85, "pad": 2},
    )
    return fig, ax


fig, ax = make_map(
    map_extent(footprints),
    f"HyPlan flight lines and TAT-C observation swath\nPhoenix, Arizona | {AIRCRAFT_NAME} | {ALTITUDE_M / 1000:.1f} km",
    MAP_ZOOM,
)
# The area observed over the flight: the valid footprints, dissolved.
compute_airborne_ground_track(aircraft).plot(
    ax=ax,
    facecolor="tab:blue",
    edgecolor="tab:blue",
    alpha=0.28,
    linewidth=0.6,
    transform=GEOGRAPHIC,
    zorder=2,
)
ax.plot(
    trajectory["longitude"],
    trajectory["latitude"],
    color="black",
    linewidth=1.2,
    transform=GEOGRAPHIC,
    zorder=3,
)
for index, marker, color in ((0, "^", "tab:green"), (-1, "s", "tab:orange")):
    point = trajectory.iloc[index]
    ax.plot(
        point["longitude"],
        point["latitude"],
        marker=marker,
        color=color,
        markeredgecolor="white",
        markersize=8,
        transform=GEOGRAPHIC,
        zorder=5,
    )
ax.legend(
    handles=[
        Line2D([], [], color="black", label="Flight path"),
        Patch(facecolor="tab:blue", alpha=0.35, label="TAT-C swath (with attitude)"),
        Line2D([], [], marker="^", color="tab:green", linestyle="none", label="Start"),
        Line2D([], [], marker="s", color="tab:orange", linestyle="none", label="End"),
    ],
    loc="upper right",
    fontsize=9,
    framealpha=0.95,
)
plt.show()
```

## Banking Shifts the Swath

For a sensor fixed to the aircraft's body, banking changes the scan's ground
intersection. The blue footprints use the estimated roll and pitch; the red
outlines show level flight, from a trajectory read from the same table without its
attitude columns (without them, the aircraft is level). HyPlan's connector can contain turns in both
directions, and the estimated roll follows those changes rather than forcing one
continuous bank.

The plot includes 45 seconds before and after the first connection. Straight cutoffs
at the plotted window boundary do not indicate a sudden physical roll transition.

```python
level_aircraft = Aircraft(
    name=AIRCRAFT_NAME,
    trajectory=Trajectory.from_dataframe(trajectory.drop(columns=["heading", "roll", "pitch"])),
    instruments=[scanner],
)
level = collect_airborne_ground_track(level_aircraft)

first_connection = turn_segments[0]
connection = np.flatnonzero(trajectory["segment_name"].eq(first_connection))
window = np.arange(
    max(connection[0] - 45, 0), min(connection[-1] + 45, len(trajectory) - 1) + 1
)

turn_footprints = footprints.iloc[window]
turn_level = level.iloc[window]
fig, ax = make_map(
    map_extent(turn_footprints, turn_level, padding=0.12),
    "First turn: aircraft attitude shifts the observed ground",
    TURN_MAP_ZOOM,
)
turn_footprints[["geometry"]].dissolve().plot(
    ax=ax,
    facecolor="tab:blue",
    edgecolor="tab:blue",
    alpha=0.28,
    linewidth=0.6,
    transform=GEOGRAPHIC,
    zorder=2,
)
turn_level.iloc[::10].plot(
    ax=ax,
    facecolor="none",
    edgecolor="tab:red",
    linewidth=0.8,
    alpha=0.8,
    transform=GEOGRAPHIC,
    zorder=3,
)
ax.plot(
    trajectory["longitude"].iloc[window],
    trajectory["latitude"].iloc[window],
    color="black",
    linewidth=1.3,
    transform=GEOGRAPHIC,
    zorder=4,
)
ax.legend(
    handles=[
        Line2D([], [], color="black", label="Flight path"),
        Patch(facecolor="tab:blue", alpha=0.35, label="With roll/pitch"),
        Patch(
            facecolor="none",
            edgecolor="tab:red",
            label="Level flight (every 10th sample)",
        ),
    ],
    loc="upper right",
    fontsize=9,
    framealpha=0.95,
)
plt.show()
```

## Observe a Level Above the Ground

By default the footprints are on the WGS 84 ellipsoid (`elevation=0`). The `elevation` argument
moves the observed level up, for example to the height of a layer in a storm: the aircraft stays
at its altitude, and the scan rays meet a surface `elevation` meters above the ellipsoid, in the
same reference as the altitudes. A downward-looking scanner therefore sees a narrower band at a
higher level, `2 (h - z) tan(half angle)` wide in straight, level flight at height `h` above the
level `z`, and nothing at or above the aircraft: those samples are flagged with
`valid_obs = False`.

The band is the **geometric** overlap of the scan with that level. Whether the sensor detects
anything there (its sensitivity, or attenuation along the beam) is not modeled. The elevation is
one constant height for the whole flight; it does not follow the terrain or the top of a cloud.

The levels below are 0, 3, and 5 km, and 6 km, which is above the aircraft. The table compares
the width on the straight, level leg used in the checks above with the formula.

```python
# Observed levels in meters above the WGS 84 ellipsoid, like the altitudes. The aircraft flies
# at ALTITUDE_M, so the last level is above it and cannot be observed by a downward scan.
OBSERVED_LEVELS_M = [0, 3000, 5000, 6000]


def observe_level(level_m):
    """Footprints at an observed level, and a summary row with the width on a level leg."""
    at_level = collect_airborne_ground_track(aircraft, elevation=level_m)
    is_valid = at_level["valid_obs"].to_numpy()
    # the interior of the straight, level leg used in the checks above
    heights = trajectory["altitude"].to_numpy()[leg] - level_m
    both = heights > 0
    edge_lon, edge_lat = get_edges(level_m)
    widths = GEOD.inv(
        edge_lon[leg[both], 0],
        edge_lat[leg[both], 0],
        edge_lon[leg[both], 1],
        edge_lat[leg[both], 1],
    )[2]
    area = (
        at_level.loc[is_valid, ["geometry"]].dissolve().to_crs(projected_crs).area.sum() / 1e6
        if is_valid.any()
        else 0.0
    )
    row = {
        "observed level (km)": level_m / 1000,
        "valid footprints": f"{is_valid.sum()} of {len(at_level)}",
        "width on a level leg (km)": np.median(widths) / 1000 if both.any() else np.nan,
        "2 (h - z) tan(half angle) (km)": (
            np.median(2 * heights * np.tan(np.radians(HALF_ANGLE_DEG))) / 1000
            if (heights > 0).all()
            else np.nan
        ),
        "observed area (km²)": area,
    }
    return at_level, row


observed = {level_m: observe_level(level_m) for level_m in OBSERVED_LEVELS_M}
pd.DataFrame([row for _, row in observed.values()]).set_index("observed level (km)").round(2)
```

```python
LEVEL_COLORS = {0: "#9db7d6", 3000: "#2465a3", 5000: "#0b2e59"}


def plot_side_view():
    """The scan across the track of a level flight, and the band at each level."""
    h_km = ALTITUDE_M / 1000

    def half_width_km(z_km):
        return (h_km - z_km) * np.tan(np.radians(HALF_ANGLE_DEG))

    fig, ax = plt.subplots(figsize=(8, 3.4))
    fan = [-half_width_km(0), 0, half_width_km(0)]
    ax.fill(fan, [0, h_km, 0], color="tab:blue", alpha=0.10, linewidth=0)
    ax.plot(fan, [0, h_km, 0], color="tab:blue", linewidth=1)
    for level_m, color in LEVEL_COLORS.items():
        z_km, w_km = level_m / 1000, half_width_km(level_m / 1000)
        ax.plot([-w_km, w_km], [z_km, z_km], color=color, linewidth=4, solid_capstyle="butt")
        ax.text(w_km + 0.2, z_km, f"{z_km:.0f} km: {2 * w_km:.1f} km wide", va="center", fontsize=9)
    top_km = OBSERVED_LEVELS_M[-1] / 1000
    ax.axhline(top_km, color="gray", linestyle="--", linewidth=1)
    ax.text(
        -half_width_km(0),
        top_km + 0.15,
        f"{top_km:.0f} km: above the aircraft, not observed",
        fontsize=9,
        color="gray",
    )
    ax.plot(0, h_km, marker="v", color="black", markersize=9)
    ax.text(0.3, h_km - 0.25, f"{AIRCRAFT_NAME} at {h_km:.1f} km", fontsize=9, va="center")
    ax.set(
        xlabel="Distance across the track (km)",
        ylabel="Height above the ellipsoid (km)",
        xlim=(-half_width_km(0) - 0.5, half_width_km(0) + 4.5),
        ylim=(-0.5, top_km + 0.8),
    )
    ax.set_title(
        f"Scan of a level flight, half angle {HALF_ANGLE_DEG:.1f} deg: the band narrows with height",
        fontsize=11,
    )
    ax.grid(alpha=0.25)
    plt.show()


def plot_turn_bands():
    """The bands observed at each level around the first turn, widest first."""
    fig, ax = make_map(
        map_extent(observed[0][0].iloc[window], padding=0.12),
        "First turn: the band observed at 0, 3, and 5 km",
        TURN_MAP_ZOOM,
    )
    for level_m, color in LEVEL_COLORS.items():
        part = observed[level_m][0].iloc[window]
        part.loc[part["valid_obs"], ["geometry"]].dissolve().plot(
            ax=ax,
            facecolor=color,
            edgecolor=color,
            alpha=0.75,
            linewidth=0.4,
            transform=GEOGRAPHIC,
            zorder=2,
        )
    ax.plot(
        trajectory["longitude"].iloc[window],
        trajectory["latitude"].iloc[window],
        color="black",
        linewidth=1.3,
        transform=GEOGRAPHIC,
        zorder=4,
    )
    ax.legend(
        handles=[Line2D([], [], color="black", label="Flight path")]
        + [
            Patch(facecolor=color, alpha=0.75, label=f"Observed at {level_m / 1000:.0f} km")
            for level_m, color in LEVEL_COLORS.items()
        ],
        loc="upper right",
        fontsize=9,
        framealpha=0.95,
    )
    plt.show()


plot_side_view()
plot_turn_bands()
```

## Animate the Flight and Growing Swath

The aircraft marker follows the same trajectory used in the analysis. The animation
advances in discrete time steps of 60 seconds (`TIME_STEP_S`): orange highlights the
one-second footprints observed during the latest step, and blue footprints accumulate
from the earlier steps. Roll and pitch are already reflected in the calculated
footprints, so the swath moves to the outside of a banked turn. The background is a
line drawing of the roads, rivers, and urban areas of Natural Earth, which keeps the
animation small.

Each frame ends one time step, so **every valid one-second footprint is included**,
either in the latest step or in the accumulated swath.
The animation is an animated WebP image embedded in the notebook, which needs no video
encoder or JavaScript and stores only the changes from one frame to the next. It plays
in a loop and holds the last frame, the complete swath, for a few seconds. Where the
animated image cannot be displayed, the last frame appears as a still image.

```python
import base64
from io import BytesIO

from IPython.display import display
from matplotlib.collections import PolyCollection
from matplotlib.path import Path as MarkerPath
from matplotlib.transforms import Affine2D
from PIL import Image

TIME_STEP_S = 60  # seconds of flight per animation frame
FRAME_MS, LAST_FRAME_MS = 200, 3000  # display time of each frame, and of the last one

fig, ax = make_map(
    map_extent(footprints), "Flight and accumulated observation swath", MAP_ZOOM, "lines"
)
fig.set_size_inches(7.2, 7.2)
fig.set_dpi(72)
fig.subplots_adjust(left=0.12, right=0.98, bottom=0.08, top=0.88)

# Project each footprint once; updating the animation only changes artists.
projected = footprints.to_crs(ax.projection.proj4_init)
valid = footprints["valid_obs"].to_numpy() & ~projected.geometry.is_empty.to_numpy()
vertices = [
    np.asarray(geometry.exterior.coords)[:, :2] if usable else np.empty((0, 2))
    for geometry, usable in zip(projected.geometry, valid)
]
accumulated = PolyCollection([], facecolors="tab:blue", edgecolors="none", alpha=0.3, zorder=2)
current = PolyCollection([], facecolors="tab:orange", edgecolors="none", zorder=4)
ax.add_collection(accumulated)
ax.add_collection(current)

longitude = trajectory["longitude"].to_numpy()
latitude = trajectory["latitude"].to_numpy()
heading = trajectory["heading"].to_numpy()
ax.plot(longitude, latitude, color="gray", linewidth=0.8, linestyle="--", transform=GEOGRAPHIC)
(flown,) = ax.plot([], [], color="black", linewidth=1.1, transform=GEOGRAPHIC, zorder=3)
# A north-pointing aircraft silhouette, rotated clockwise with its heading.
aircraft_shape = MarkerPath(
    [(0, 1), (0.12, 0.2), (0.85, -0.25), (0.85, -0.42), (0.12, -0.2), (0.1, -0.7),
     (0.36, -0.85), (0.36, -1), (0, -0.85), (-0.36, -1), (-0.36, -0.85), (-0.1, -0.7),
     (-0.12, -0.2), (-0.85, -0.42), (-0.85, -0.25), (-0.12, 0.2), (0, 1)]
)
(marker,) = ax.plot([], [], linestyle="none", marker=aircraft_shape, markersize=18,
                    color="black", transform=GEOGRAPHIC, zorder=6)
status = ax.text(0.015, 0.985, "", transform=ax.transAxes, va="top", fontsize=8, zorder=8,
                 bbox={"facecolor": "white", "edgecolor": "none", "alpha": 0.92, "pad": 4})
ax.legend(
    handles=[
        Line2D([], [], color="black", label="Aircraft path"),
        Patch(facecolor="tab:orange", label=f"Last {TIME_STEP_S} s"),
        Patch(facecolor="tab:blue", alpha=0.4, label="Accumulated swath"),
    ],
    loc="upper right",
    fontsize=8,
    framealpha=0.95,
)

# One frame at the end of each time step: the number of samples up to and including
# that time. The last step ends at or after the last sample, so every footprint is shown.
elapsed = (trajectory["time"] - trajectory["time"].iloc[0]).dt.total_seconds().to_numpy()
step_ends = np.arange(0, elapsed[-1] + TIME_STEP_S, TIME_STEP_S)
frame_counts = np.unique(np.searchsorted(elapsed, step_ends, side="right"))
valid_counts = np.r_[0, np.cumsum(valid)]


def animate_swath(frame):
    """Draw the state at the end of a time step."""
    start = frame_counts[frame - 1] if frame else 0
    count = frame_counts[frame]
    index = count - 1  # the last sample of the time step
    accumulated.set_verts([vertices[i] for i in range(start) if valid[i]])
    current.set_verts([vertices[i] for i in range(start, count) if valid[i]])
    flown.set_data(longitude[: index + 1], latitude[: index + 1])
    marker.set_data([longitude[index]], [latitude[index]])
    marker.set_marker(aircraft_shape.transformed(Affine2D().rotate_deg(-heading[index])))
    status.set_text(
        f"{trajectory['time'].iloc[index]:%H:%M:%S} UTC\n"
        f"Observed: {valid_counts[count]} / {valid.sum()}\n"
        f"Roll: {trajectory['roll'].iloc[index]:+.1f} deg"
    )
    return accumulated, current, flown, marker, status


def render_frame(frame):
    """The figure at the end of a time step, as an image."""
    animate_swath(frame)
    fig.canvas.draw()
    return Image.fromarray(np.asarray(fig.canvas.buffer_rgba())).convert("RGB")


def encode(image, **kwargs):
    """An image file, encoded as base64 text to embed in the notebook."""
    buffer = BytesIO()
    image.save(buffer, **kwargs)
    return base64.b64encode(buffer.getvalue()).decode()


frames = [render_frame(frame) for frame in range(len(frame_counts))]
plt.close(fig)
webp = encode(
    frames[0],
    format="WEBP",
    save_all=True,
    append_images=frames[1:],
    duration=[FRAME_MS] * (len(frames) - 1) + [LAST_FRAME_MS],
    loop=0,
    quality=75,
    method=6,  # the slowest and smallest encoding
)
# An animated image in HTML, with the last frame as a still image for viewers without HTML.
display(
    {
        "text/html": f'<img src="data:image/webp;base64,{webp}" '
        'alt="Animation of the flight and accumulated observation swath">',
        "image/jpeg": encode(frames[-1], format="JPEG", quality=75, optimize=True),
    },
    raw=True,
)
```
